```python
top_batsman = deliveries.groupby('batsman')['batsman_runs'].sum().reset_index()
display(top_batsman.sort_values(by='batsman_runs', ascending=False).head(5))
```

```python
dhoni_runs = deliveries.query("batsman == 'MS Dhoni'")['batsman_runs'].sum()
print(f"Total runs scored by MS Dhoni: {dhoni_runs}")
```

```python
rohit_sharma_ball_faced =deliveries.query("batsman =='RG Sharma'").shape[0]
print(f"Total balls faced by Rohit Sharma: {rohit_sharma_ball_faced}")
```

```python
gayle_sixes = deliveries.query("batsman == 'CH Gayle' and batsman_runs == 6").shape[0]
print(f"Total sixes hit by Chris Gayle: {gayle_sixes}")
```

```python
most_balls = deliveries['bowler'].value_counts().head()
print("Most balls bowled by a bowler:")
print(most_balls)
```

```python
boundaries = deliveries.query("batsman_runs == 4 or batsman_runs == 6")['batsman'].value_counts().head(1)
display(boundaries)
```

```python
valid = deliveries.query("dismissal_kind != 'run out'").groupby('bowler')['dismissal_kind'].count().reset_index(name='wickets')
display(valid.sort_values(by='wickets', ascending=False).head(5))
```
