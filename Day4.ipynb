```python
import pandas as pd

# ✅ Load CSV (using forward slashes so no escape sequence issues)

# Load datasets
matches = pd.read_csv("IPL Dataset/matches.csv")
deliveries = pd.read_csv("IPL Dataset/deliveries.csv")

# ✅ Make sure all columns show in one line (no wrapping with \)
pd.set_option("display.max_columns", 600)
pd.set_option("display.width", 600)
```

Kohli's total runs , balls and strike rate

```python
kohli = deliveries.query("batsman == 'V Kohli' ")
runs = kohli["batsman_runs"].sum()
balls = kohli.shape[0]
strike_rate = (runs / balls) * 100
print(f"Virat Kohli's Strike Rate: {strike_rate:.2f}")
```

```python
batsman_stats = deliveries.groupby("batsman").agg(runs=("batsman_runs", "sum"), balls=("batsman_runs", "count"))
batsman_stats["strike_rate"] = (batsman_stats["runs"] / batsman_stats["balls"]) * 100
qualified_batsmen = batsman_stats[batsman_stats["balls"] >= 3000]
top_10_strike_rate = qualified_batsmen.sort_values(by="strike_rate", ascending=False).head(10)
print(top_10_strike_rate)
```

```python
bowler_stats = deliveries.groupby("bowler").agg(
    runs=('total_runs','sum'),
    balls=('bowler','count')
)
bowler_stats['economy'] = (bowler_stats['runs'] / bowler_stats['balls']) * 6
qualified = bowler_stats[bowler_stats['balls'] >= 300]
print(qualified.sort_values(by='economy').head(1))
```

```python
sixes = deliveries.query("batsman_runs == 6")
print(sixes['bowler'].value_counts().head(10))
```

```python
powerplay = deliveries.query("over<=6")
pp_runs = powerplay.groupby("batsman")["batsman_runs"].sum().sort_values(ascending =False ).head(10)
display(pp_runs)
```

```python
death_overs = deliveries.query("over>=16")
death_runs = death_overs.groupby("batsman")["batsman_runs"].sum().sort_values(ascending =False ).head(10)
display(death_runs)
```
